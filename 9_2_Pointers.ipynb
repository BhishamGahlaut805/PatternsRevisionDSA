# PATTERN 9 2 Pointers

2Pointers is a very important pattern in DSA. It is used to solve problems related to arrays and linked lists. The idea is to use two pointers to traverse the data structure and find the solution.

1.2 Sum on Sorted Array : Given an array of integers nums sorted in non-decreasing order, find two numbers such that they add up to a specific target number. Return the indices of the two numbers (1-indexed) as an integer array answer of size 2, where 1 <= answer[0] < answer[1] <= nums.length.

approach using 2 pointers :
    a. Initialize two pointers, left and right, at the beginning and end of the array respectively.
    b. While left < right:
        i. Calculate the sum of the elements at the left and right pointers.
        ii. If the sum is equal to the target, return the indices of the two numbers.
        iii. If the sum is less than the target, increment the left pointer to increase the sum.
        iv. If the sum is greater than the target, decrement the right pointer to decrease the sum.

2. 3Sum Problem : Given an array , find all unique triplets in the array which gives the sum of zero.
approach using 2 pointers :
    a. Sort the array to make it easier to avoid duplicates.
    b. Iterate through the array with an index i, treating nums[i] as the first element of the triplet.
    c. For each nums[i], initialize two pointers: left at i + 1 and right at the end of the array.
    d. While left < right:
        i. Calculate the sum of nums[i], nums[left], and nums[right].
        ii. If the sum is zero, add the triplet to the result list and move both pointers to find other potential triplets.
        iii. If the sum is less than zero, increment the left pointer to increase the sum.
        iv. If the sum is greater than zero, decrement the right pointer to decrease the sum.
    e. Skip duplicate values for nums[i] to ensure unique triplets.
TC: O(n^2) for 3Sum, O(n) for 2 Sum

3. 4Sum Problem : Given an array, find all unique quadruplets in the array which gives the sum of a target value.
approach using 2 pointers :
    a. Sort the array to make it easier to avoid duplicates.
    b. Iterate through the array with an index i, treating nums[i] as the first element of the quadruplet.
    c. For each nums[i], iterate through the array with an index j starting from i + 1, treating nums[j] as the second element of the quadruplet.
    d. Initialize two pointers: left at j + 1 and right at the end of the array.
    e. While left < right:
        i. Calculate the sum of nums[i], nums[j], nums[left], and nums[right].
        ii. If the sum is equal to the target, add the quadruplet to the result list and move both pointers to find other potential quadruplets.
        iii. If the sum is less than the target, increment the left pointer to increase the sum.
        iv. If the sum is greater than the target, decrement the right pointer to decrease the sum.
    f. Skip duplicate values for nums[i] and nums[j] to ensure unique quadruplets.
TC: O(n^3) for 4Sum, O(n^2) for 3Sum, O(n) for 2 Sum
