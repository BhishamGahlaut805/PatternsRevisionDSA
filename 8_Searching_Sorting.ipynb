# PATTERN 8 Searching And Sorting Algorithms

Binary Search is a searching algorithm that finds the position of a target value within a sorted array. It compares the target value to the middle element of the array. If they are not equal, the half in which the target cannot lie is eliminated and the search continues on the remaining half until it is successful.

1.First Bad Version : Given an array of n elements, find the first bad version. You can call a function isBadVersion(version) which will return whether version is bad. Implement a function to find the first bad version.
approach using Binary Search:
    a. Initialize two pointers, left and right, to the start and end of the array respectively.
    b. While left is less than right, calculate the middle index mid as (left + right) / 2.
    c. If isBadVersion(mid) returns true, it means the first bad version is in the left half of the array, so update right to mid.
    d. If isBadVersion(mid) returns false, it means the first bad version is in the right half of the array, so update left to mid + 1.
    e. Repeat steps b to d until left is equal to right. At this point, left (or right) will be pointing to the first bad version.
    f. Return left (or right) as the index of the first bad version.
    TC: O(log n) because we are dividing the search space in half with each iteration.

3.Find Minimum Element in Rotated Sorted Array : Given a sorted array that has been rotated at some pivot unknown to you beforehand, find the minimum element in the array.
    a. Initialize two pointers, left and right, to the start and end of the array respectively.
    b. While left is less than right, calculate mid as (left + right) / 2.
    c. If nums[mid] is greater than nums[right], it means the minimum element is in the right half of the array, so update left to mid + 1.
    d. If nums[mid] is less than or equal to nums[right], it means the minimum element is in the left half of the array, so update right to mid.
    e. Repeat steps b to d until left is equal to right. At this point, left (or right) will be pointing to the minimum element.
    f. Return nums[left] (or nums[right]) as the minimum element in the rotated sorted array.
    TC: O(log n) because we are dividing the search space in half with each iteration.

Technique : Binary Search on Solution Space : This technique is used when the solution space is not explicitly given but can be inferred from the problem constraints. It involves applying binary search on the range of possible solutions to find the optimal solution.
approach :
    a.Find the minium and maximum possible values for the solution space based on the problem constraints.
    b.While the minimum value is less than or equal to the maximum value, calculate mid as (min + max) / 2.
    c. Check if mid is a valid solution based on the problem constraints. If it is, update the maximum value to mid - 1 to search for a potentially better solution. If it is not valid, update the minimum value to mid + 1 to search for a valid solution.
d. Repeat steps b and c until the minimum value is greater than the maximum value. At this point, the minimum value will be pointing to the optimal solution.
    e. Return the minimum value as the optimal solution.
    TC: O(log n) because we are dividing the search space in half with each iteration.
