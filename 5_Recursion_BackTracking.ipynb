# PATTERN 5 Recursion and BackTracking

Recursion is a process in which a function calls itself directly or indirectly. The corresponding function is called a recursive function. Recursion is used to solve problems that can be broken down into smaller, similar subproblems.

Backtracking is a general algorithm for finding all (or some) solutions to computational problems, notably constraint satisfaction problems, that incrementally builds candidates to the solutions, and abandons a candidate ("backtracks") as soon as it determines that the candidate cannot possibly be completed to a valid solution.

```python
1.Tiling Problem : Given a 2 x n board and tiles of size 2 x 1, count the number of ways to tile the given board using the 2 x 1 tiles. A tile can either be placed horizontally (in a single row) or vertically (covering both rows).

approach using recursion :
    a. If we place the tile vertically, then we are left with a 2 x (n-1) board to tile.
    b. If we place the tile horizontally, then we must place another tile horizontally to cover the second row, and we are left with a 2 x (n-2) board to tile.
    c. Therefore, the total number of ways to tile the board is the sum of the number of ways to tile a 2 x (n-1) board and a 2 x (n-2) board.
    d.Base cases: If n = 0, there is 1 way to tile the board (by not placing any tiles). If n = 1, there is 1 way to tile the board (by placing one tile vertically).
    e. The recursive formula can be expressed as:
    T(n) = T(n-1) + T(n-2)
TC: O(2^n) - The time complexity is exponential because we are making two recursive calls for each value of n.
```

2.Friend's Pairing Problem :  Statement: Given n friends, each one can remain single or can be paired up with some other friend. Each friend can be paired only once. Find out the total number of ways in which friends can remain single or can be paired up.

Recurrence Relation: Let f(n) be the number of ways in which n friends can remain single or can be paired up. Then, we can derive the following recurrence relation:
f(n) = f(n-1) + (n-1) * f(n-2)
Explanation: The first term, f(n-1), represents the case where the nth friend remains single. The second term, (n-1) * f(n-2), represents the case where the nth friend is paired with any of the (n-1) other friends, and then we recursively solve for the remaining friends.

Base Cases:
f(0) = 1 (There is one way to pair zero friends)
f(1) = 1 (There is one way to pair one friend, which is to remain single)

TC:(using DP) O(n) - The time complexity is linear because we compute the value of f(n) using previously computed values of f(n-1) and f(n-2).

3.Tower of Hanoi  :
Threes steps to solve the Tower of Hanoi problem:
1. Move n-1 disks from source rod to auxiliary rod using destination rod.
2. Move the nth disk from source rod to destination rod.
3. Move the n-1 disks from auxiliary rod to destination rod using source rod.
TC: O(2^n) - The time complexity is exponential because each disk requires two recursive calls, leading to a total of 2^n moves.

## Backtracking

Backtracking is a powerful algorithmic technique for solving problems that involve searching through a large solution space. It is particularly useful for problems where we need to explore all possible combinations or arrangements of a set of elements, such as generating permutations, combinations, or solving constraint satisfaction problems.

2.N-Queens Problem:
The N-Queens problem is a classic backtracking problem that involves placing N queens on an N x N chessboard such that no two queens threaten each other. The goal is to find all possible arrangements of the queens on the board.
Approach:
Function checkSafe(board, row, col, N):
    for i from 0 to row-1:
        if board[i][col] == 1:
            return false
    for i, j in (row-1, col-1) to (0, 0):
        if board[i][j] == 1:
            return false
    for i, j in (row-1, col+1) to (0, N-1):
        if board[i][j] == 1:
            return false
    return true
Function solveNQueens(board, row, N):
    if row == N:
        print(board)
        return
    for col from 0 to N-1:
        if checkSafe(board, row, col, N):
            board[row][col] = 1
            solveNQueens(board, row + 1, N)
            board[row][col] = 0 // backtrack
