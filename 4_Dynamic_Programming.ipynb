# PATTERN 4 Dynamic Programming

Dynamic programming (DP) is a powerful technique used to solve problems by breaking them down into simpler subproblems and storing the results of these subproblems to avoid redundant computations. It is particularly useful for optimization problems and problems with overlapping subproblems and optimal substructure properties.

````python
1.Climibing Stairs:
    a. The climbing stairs problem involves finding the number of distinct ways to reach the top of a staircase with n steps, where you can take either 1 or 2 steps at a time.
    b. To solve this problem using dynamic programming, you can use an array dp where dp[i] represents the number of ways to reach step i.
    c. The recurrence relation is: dp[i] = dp[i-1] + dp[i-2], with base cases dp[0] = 1 and dp[1] = 1.
    d. Iterate from step 2 to n, filling in the dp array based on the recurrence relation.
    e. The final answer will be stored in dp[n].
TC:O(N)
````

````python
5.Binary KnapSack Problem : M denoting the KnapSack capacity and N denoting the number of items. Each item has a weight and a value. The goal is to maximize the total value of items that can be included in the knapsack without exceeding its capacity.

a.Memoization (top-down approach):
    i.Define a 2D DP[n+1][m+1] array where n is the number of items and m is the capacity of the knapsack.
    ii.Initialize the DP array with -1 to indicate uncomputed states.
    iii. Create a recursive function that takes the current item index and remaining capacity as parameters.
    iv. Base Case : if(n == 0 || m == 0) return 0; // No items or no capacity
    v. Check if the value is already computed in DP[n][m] !=-1. If yes, return that value.
    vi.call recursively for 2 cases below :
        Include Case : call recursively ( value[n] + function(n-1, m-weight[n-1]) + value[n-1] )// Include the current item
        Exclude Case : call recursively (0 + function(n-1, m)) // Exclude the current item
    vii. Store the maximum of the two cases in DP[n][m] and return that value.

b.Top-Down Approach (Iterative):
    i. Define a 2D DP[n+1][m+1] array where n is the number of items and m is the capacity of the knapsack.
    ii. Initialize the first row and first column of the DP array to 0, as they represent the base cases of no items or no capacity.
    iii. Iterate through each item (i from 1 to n) and each capacity (j from 1 to m):
        - If the weight of the current item is less than or equal to the current capacity (weight[i-1] <= j):
            - DP[i][j] = max(value[i-1] + DP[i-1][j-weight[i-1]], DP[i-1][j]) // Include or exclude the item
        - Else:
            - DP[i][j] = DP[i-1][j] // Exclude the item
    iv. The maximum value that can be obtained is stored in DP[n][m].
TIME COMPLEXITY : O(N*M) where N is the number of items and M is the capacity of the knapsack.
SPACE COMPLEXITY : O(N*M) for the DP array.
````

````python
6.Edit Distance Problem: Given two strings, the edit distance is the minimum number of operations required to convert one string into the other. The allowed operations are insertion, deletion, and substitution of a single character.

Given two strings `str1` and `str2`, the edit distance can be computed using a 2D DP table where `dp[i][j]` represents the edit distance between the first `i` characters of `str1` and the first `j` characters of `str2`.

Using Bottom-up 2D DP (size `(m+1) x (n+1)` where `m` is the length of `str1` and `n` is the length of `str2`):
    a.base Case for (j = 0): dp[i][0] = i (deleting all characters from str1)
    b.base Case for (i = 0): dp[0][j] = j (inserting all characters of str2 into str1)
    c.Recurrence Relation:
    d.  If str1[i-1] == str2[j-1], then dp[i][j] = dp[i-1][j-1] (no operation needed)
    e.  If str1[i-1] != str2[j-1], then
        dp[i][j] = 1 + min(dp[i-1][j],    // Deletion
                           dp[i][j-1],    // Insertion
                           dp[i-1][j-1])  // Substitution
    f.return dp[m][n] as the final answer.
Time Complexity: O(m*n)
````

````python
7.Coin change problem: Given a set of coin denominations and a target amount, determine the number of ways to make change for that amount using the given denominations.

Using Recursive memoization approach : We define a 2D DP of Size (n+1) x (amount+1) where n is the number of coin denominations and amount is the target amount. The DP table will store the number of ways to make change for each amount using the available denominations.
a.Base Case : if(amount == 0) return 1; // There is one way to make change for amount 0 (using no coins)
b.Base Case : if(n == 0) return 0; // If there are no coin denominations, there are no ways to make change for any positive amount
c.Recursive Relation : dp[i][j] = dp[i-1][j] + dp[i][j - coin[i-1]]; // The number of ways to make change for amount j using the first i coins is the sum of the ways to make change without using the ith coin and the ways to make change using the ith coin.
d.return dp[n][amount]; // The final answer will be stored in dp[n][amount], which represents the number of ways to make change for the target amount using all available coin denominations.

It is Similar to the 0/1 Knapsack problem, but instead of maximizing value, we are counting the number of ways to achieve a target amount.
Time Complexity: O(n * amount), where n is the number of coin denominations and amount is the target amount. This is because we fill a DP table of size (n+1) x (amount+1).
````

````python
8.Parition Problem: Given a set of integers, determine if it can be partitioned into two subsets such that the sum of elements in both subsets is equal.
Approach :
    a.Define a boolean DP array `dp[i][j]` where `i` represents the number of elements considered and `j` represents the target sum. The value of `dp[i][j]` will be true if a subset with sum `j` can be formed using the first `i` elements.
    b.Initialize the DP array with base cases: `dp[0][0] = true` (an empty subset can form sum 0) and `dp[i][0] = true` for all `i` (sum 0 can always be formed with an empty subset).
    c.Fill the DP array using the following recurrence relation:
    ```
    dp[i][j] = dp[i-1][j] || dp[i-1][j - arr[i-1]] if j >= arr[i-1]
    dp[i][j] = dp[i-1][j] if j < arr[i-1]
    ```
    d.Return the value of `dp[n][target]`, where `n` is the number of elements in the set and `target` is half of the total sum of the elements (if the total sum is odd, return false).
Time complexity: O(n * target), where n is the number of elements and target is half of the total sum.
````

````python
11.Decode ways : Given a string containing digits from 0-9, determine the total number of ways to decode it. The mapping is as follows: 'A' -> 1, 'B' -> 2, ..., 'Z' -> 26. For example, the input "12" can be decoded as "AB" (1 2) or "L" (12), so the output should be 2.
Approach :
    a.Initialize a 1D DP array of size n+1, where n is the length of the input string. The DP array will store the number of ways to decode the substring up to each index.
    b.Set dp[0] = 1, as there is one way to decode an empty string.
    c.Set dp[1] = 1 if the first character is not '0', otherwise set dp[1] = 0, as '0' cannot be decoded.
    d.for i from 2 to n:
        i. If the character at index i-1 is not '0', add dp[i-1] to dp[i].
        ii. If the two characters at indices i-2 and i-1 form a valid number between 10 and 26, add dp[i-2] to dp[i].
            we can check using stoi(s.substr(i-2, 2)) >= 10 && stoi(s.substr(i-2, 2)) <= 26 for 2 digit number.
            we can also check using s[i-2] == '1' || (s[i-2] == '2' && s[i-1] <= '6') for 2 digit number.
            for 1 digit number we can check using s[i-1] != '0' and s[i-1]-'0' >= 1 and s[i-1]-'0' <= 9.
    e. Return dp[n] as the total number of ways to decode the entire string.

Time complexity: O(n), where n is the length of the input string, as we are iterating through the string once.
Space complexity: O(n), as we are using a DP array of size n+1 to store the number of ways to decode each substring.
````

````python
12.Unique Paths : Given a grid of size m x n, you are initially positioned at the top-left corner (i.e., grid[0][0]). You can only move either down or right at any point in time. Your goal is to reach the bottom-right corner (i.e., grid[m-1][n-1]). How many possible unique paths are there?
Approach :
    a.Initialize a 2D array dp of size m x n, where dp[i][j] represents the number of unique paths to reach cell (i, j).
    b.Set dp[0][0] = 1, as there is only one way to be at the starting position.
    c.Iterate through the grid and fill the dp array using the relation:
       dp[i][j] = dp[i-1][j] + dp[i][j-1]
    d.return dp[m-1][n-1] as the final answer.

TC:O(N^2) SC:O(N^2)
````

````python
13.Kadane's Algorithm (Maximum Subarray Sum) :
    a.Problem Statement: Given an integer array, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.
    b. Approach:
        i.Take the two variables currentSum and maximumSum. Initialize currentSum to 0 and maximumSum to the smallest possible integer value.
        ii.Iterate through each element of the array:
            - Add the current element to currentSum.
            - If currentSum is greater than maximumSum, update maximumSum.
            - If currentSum becomes negative, reset it to 0 (as a negative sum will not contribute to a maximum sum).
        iii. Return maximumSum after iterating through the array.
Time complexity: O(n), where n is the number of elements in the array.
Space complexity: O(1), as we are using a constant amount of space.
````
