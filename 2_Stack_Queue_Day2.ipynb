```python
1.Array of implementation of Stack :
Requires an array and a pointer which points to the top of the Stack :
    int *arr;
    int top;
    int size;
    With a constructor for initializing :
    Stack(n):
        arr=new int[n];
        top=-1;
        size=n;
Operations :
1.Push element : Takes O(1) Push at the top of the Stack and increment the Top
2.Peek : Takes O(1) Returns the Top element of the Stack
3.Pop: Takes O(1) Removes element from the top of the stack and decrements the top
4.isEmpty : check where the top==-1 means Stack is empty
5.isFull : When size of Stack is equal to determined size n takes o(1)
```

```python
3.Solution of Prefix Expression :
    Infix = Normal Mathematical Expressions
    Prefix= Operator at starting, then operands
    PostFix= Operand at Starting, then operator , Known as reverse Polish Notation , By default Storage usage for computer systems
For A Prefix Expression such as :  "++1/*2+346*45"

Procedure for solving to return its resultant value :
    a.Maintain a Stack<int> for Storing the characters
    b.Traverse the String from the End and do (from N to 1) :
    c.If the current character is not a operator then Push that Number into the stack (not operators)
    d.If the current character is a Operator '+','-','*','/','^' etc then
    e.Operand1 = stack.top() and pop element from stack
    f.Operands2 = stack.top() and pop element from stack
    g.Perform Operand1 * operands2 and store result in stack. where * ='+','-','*','/','^'
    h.Return the top of the Stack

Return Stack.top()
TC:O(N) ; SC:O(N)
```

# QUEUE DATA STRUCTURE

A Queue is a Data Structure which supports First in First Out Property.Following types of Queue are there:
1. Simple Queue
2. Circular Queue
3. Priority Queue
4. Double Ended Queue (Deque)

```python
C++ STL Provides the following Queue Data Structure:
1. Simple Queue :
   - queue<datatype> q;
    Operations:
    - q.push() : Inserts an element at the end of the queue. TC: O(1)
    - q.pop() : Removes the front element of the queue. TC: O(1)
    - q.front() : Returns the front element of the queue. TC: O(1)
    - q.back() : Returns the last element of the queue. TC: O(1)
    - q.empty() : Returns true if the queue is empty, otherwise false. TC: O(1)
    - q.size() : Returns the number of elements in the queue. TC: O(1)

2. Circular Queue :
   - circular_queue<datatype> cq;
    Operations:
    - cq.enqueue() : Inserts an element at the end of the circular queue. TC: O(1)
    - cq.dequeue() : Removes the front element of the circular queue. TC: O(1)
    - cq.front() : Returns the front element of the circular queue. TC: O(1)
    - cq.back() : Returns the last element of the circular queue. TC: O(1)
    - cq.empty() : Returns true if the circular queue is empty, otherwise false. TC: O(1)
    - cq.size() : Returns the number of elements in the circular queue. TC: O(1)

3. Priority Queue :
   - priority_queue<datatype> pq;
    Operations:
    - pq.push() : Inserts an element into the priority queue. TC: O(log n)
    - pq.pop() : Removes the highest priority element from the priority queue. TC: O(log n)
    - pq.top() : Returns the highest priority element of the priority queue. TC: O(1)
    - pq.empty() : Returns true if the priority queue is empty, otherwise false. TC: O(1)
    - pq.size() : Returns the number of elements in the priority queue. TC: O(1)

4. Double Ended Queue (Deque) :
   - deque<datatype> dq;
    Operations:
    - dq.push_front() : Inserts an element at the front of the deque. TC: O(1)
    - dq.push_back() : Inserts an element at the back of the deque. TC: O(1)
    - dq.pop_front() : Removes the front element of the deque. TC: O(1)
    - dq.pop_back() : Removes the back element of the deque. TC: O(1)
    - dq.front() : Returns the front element of the deque. TC: O(1)
    - dq.back() : Returns the back element of the deque. TC: O(1)
    - dq.empty() : Returns true if the deque is empty, otherwise false. TC: O(1)
    - dq.size() : Returns the number of elements in the deque. TC: O(1)
```
