# PATTERN 1 - LINKED LIST

 Linked list is a linear data structure where each element is a separate object. Each element (we will call it a node) of a list is comprising of two items - the data and a reference to the next node. The last node has a reference to null. The entry point into a linked list is called the head of the list. It should be noted that head is not a separate node, but the reference to the first node. If the list is empty then the head is a null reference.

```python
Important DSA LINKED LIST Operations and Functions :
1.Traversal :TC:O(n) -> For traversing a Linked List we create a temporary pointer that references to the current head, while head is not null,print the element and increment the head pointer
2.Insertion in Linked List (Singly) :
    a.At front : TC:O(1) -> create a pointer with data of new head and make it then head
    b.At End : TC:O(n) -> Traverse whole linked list and insert at End
    c.Between nodes :TC:O(n) (Worst case) -> Traverse the before node and make new connections
    d.Insertion at Index : TC:(n) (Worst Case)-> Move to the index-1 position and then insert that element
3.Deletion Operation(Singly Linked List):
    Similar to insertions , Deletion can be done from head, from index position or end position
    TC: O(n) in Worst case
```

```python
Circular Linked List :
Similar operations as of Singly Linked List but the end node points to the Head again making a circular Loop :
1.Insertion at Head :TC:O(n) -> Must move the Head to the end of the List and then insert making New Head.
2.Insertion at Tail :TC:O(n) -> Similar as of Singly linked List
3.Insertion at Middle :TC:O(n)->Similar as of Singly linked List
4.Deletion operations are similar as of Singly linked list noting the fact the the tail must point to the Head again
```

```python
Doubly Linked List :
Node contains front and back pointers for ease of traversal and insertions, Operations performed are similar as that of Singly Linked List but taking care of the prev and next pointers to be joined appropriately
```
