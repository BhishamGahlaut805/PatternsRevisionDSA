# PATTERN 7 Sliding Window and Hash Map

Sliding Window is a technique used to solve problems that involve arrays or lists, where you need to find a subarray or substring that meets certain criteria. The idea is to maintain a "window" of elements in the array and slide it across the array to find the desired result.

1.Maximum Average Subarray I :
approach : 
    a.Compute the sum of the first k elements in the array and store it as the current maximum sum.
    b.Slide the window by removing the first element of the current window and adding the next element in the array to the current sum.
    c.Update the current maximum sum if the new sum is greater than the previous maximum sum.
    d.return the maximum average by dividing the maximum sum by k.
Key:maxSum 
Time complexity: O(n)


2.Sliding Window Maximum: Given an array of integers and a sliding window size k, find the maximum value in each window as it slides from left to right across the array.
Approach using deque :
    1.Initialize a deque to store the INDICES of the elements in the current window.
    2.Iterate through the array and for each element, remove the indices of elements that are outside the current window from the front of the deque.
    3.Remove the indices of elements that are smaller than the current element from the back of the deque, as they will not be needed for future windows.
    4.Add the index of the current element to the back of the deque.
    5.If the current index is greater than or equal to k - 1, add the element at the front of the deque (which is the maximum for the current window) to the result list.
    6.return the result list containing the maximum values for each window.
Time complexity: O(n)

3.Permutation in String: Given two strings s1 and s2, return true if s2 contains a permutation of s1, or false otherwise. In other words, return true if one of s1's permutations is the substring of s2.
Approach using sliding window and hashmap:
    a. Create a hashmap to store the frequency of characters in s1.
    b. Initialize two pointers, left and right, to represent the current window in s2
    c. Iterate through s2 using the right pointer, and for each character, decrement its frequency in the hashmap.
    d. If the frequency of the character becomes zero, remove it from the hashmap.
    e. If the size of the hashmap is zero, it means all characters in s1 are present in the current window of s2, so return true.
    f. Move the left pointer and increment the frequency of the character at the left pointer in the hashmap.
    g. If the frequency of the character becomes one, add it back to the hashmap.
    h. Continue this process until the right pointer reaches the end of s2.
    i. If no permutation of s1 is found in s2, return false.
TC: O(n) where n is the length of s2

7.Minimum swaps to group all 1s together in a circular array: Given a circular binary array, find the minimum number of swaps required to group all 1s together in the array.

    b. Initialize a variable minSwaps to store the minimum number of swaps required, and set it to the length of the array.
    c. Iterate through the array using a sliding window of size totalOnes, and for each window, count the number of 0s in the window and update minSwaps if the count of 0s is less than the current minSwaps.Traverse the array 2*length of the array to handle the circular nature.
    d. Return minSwaps as the minimum number of swaps required to group all 1s together in the circular array.
This logic of finding the minimum number of 0s in a window of size totalOnes works because the minimum number of swaps required to group all 1s together is equal to the minimum number of 0s in any window of size totalOnes. By sliding the window across the array, we can find the window with the least number of 0s, which will give us the minimum number of swaps needed. Because by counting the number of 0s in a window of size totalOnes, we can determine how many 1s are already grouped together in that window. The fewer 0s there are, the more 1s are already grouped together, and thus fewer swaps are needed to group all 1s together. Therefore, by finding the window with the least number of 0s, we can determine the minimum number of swaps required to group all 1s together in the circular array.
TC:O(n) where n is the length of the array, SC: O(1)

approach using variable size sliding window:
    a. Initialize two pointers, left and right, to represent the current window in nums.
    b. Initialize a variable currentSum to store the sum of the elements in the current window, and set it to 0.
    c. Initialize a variable minLength to store the minimum length of the subarray found so far, and set it to infinity.
    d. Iterate through nums using the right pointer, and for each element, add it to currentSum.
    e. While currentSum is greater than or equal to target, update minLength with the length of the current window (right - left + 1), and subtract nums[left] from currentSum and increment left to shrink the window from the left side.

TC: O(n) where n is the length of nums, SC: O(1)

4.Longest Repeating Character Replacement : Given a string s and an integer k, you can choose any character of the string and change it to any other uppercase English character. You can perform this operation at most k times. Return the length of the longest substring containing the same letter you can get after performing the above operations.
Approach using sliding window and hashmap:
    a.Maintain a hashmap to store the frequency of characters in the current window.
    b. Keep track of the maximum frequency of any character in the current window using a variable maxFreq.
    c. Initialize a pointer left=0 and MaxFreq=0;
    d. for right from 0 to s.length()-1:
    e. Increment the frequency of s[right] in the hashmap and update maxFreq if necessary
    f. If the size of the current window (right - left + 1) - maxFreq > k, decrement the frequency of s[left] in the hashmap and move the left pointer to the right.
    g.Answer will the current window size (right - left + 1) after the loop ends, as it represents the length of the longest substring containing the same letter after performing at most k operations.
The main idea is to maintain a sliding window of characters in the string s, and keep track of the frequency of each character in the window using a hashmap. The maximum frequency of any character in the window is also tracked using the variable maxFreq. If the size of the current window minus maxFreq is greater than k, it means that we need to change more than k characters to make all characters in the window the same, so we shrink the window from the left by decrementing the frequency of s[left] in the hashmap and moving the left pointer to the right. The final answer will be the size of the largest valid window found during this process.

1.Two Sum Problem : Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target. You may assume that each input would have exactly one solution, and you may not use the same element twice. You can return the answer in any order.
Approach using hashmap:

    a.Keep an Hashmap to store the numbers and their indices as we iterate through the array.
    b. For each number, calculate its complement (target - number).
    c. Check if the complement exists in the hashmap. If it does, return the indices of the current number and the complement.
    d. If the complement does not exist, store the current number and its index in the hashmap.
    e. Return the indices of the two numbers that add up to the target.
TC: O(n) where n is the length of nums, SC: O(n) for hashmap

2.Contains Duplicate: Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.
Approach using hashmap:
    a. Create an empty hashmap to store the frequency of each number in the array.
    b. Iterate through the array and for each number, check if it already exists in the hashmap.
    c. If it does, return true as a duplicate has been found.
    d. If it does not exist, add the number to the hashmap with a frequency of 1.
    e. If the loop completes without finding any duplicates, return false.
TC: O(n) where n is the length of nums, SC: O(n) for hashmap

3.Valid Anagram: Given two strings s and t, return true if t is an anagram of s, and false otherwise. An Anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once.
Approach using hashmap:
    a. Create a hashmap to store the frequency of characters in string s.
    b. Iterate through string t and for each character, decrement its frequency in the hashmap.
    c. If the frequency of any character becomes negative or if a character in t does not exist in the hashmap, return false as t cannot be an anagram of s.
    d. After iterating through t, check if all frequencies in the hashmap are zero. If they are, return true as t is an anagram of s; otherwise, return false.
TC:O(n) where n is the length of s and t, SC: O(1) for hashmap (since there are only 26 lowercase English letters).

4.Group Anagrams: Given an array of strings strs, group the anagrams together. You can return the answer in any order.
Approach using hashmap:
    a. Create a hashmap to store the grouped anagrams, where the key is a sorted version of the string and the value is a list of strings that are anagrams of each other.
    b. Iterate through the array of strings, and for each string, sort its characters to create the key.
    c. Add the original string to the list corresponding to the sorted key in the hashmap.
    d. After iterating through all strings, return the values of the hashmap as a list of lists, where each inner list contains grouped anagrams.
TC:O(n*m*log(m)) where n is the number of strings and m is the maximum length of a string, SC: O(n*m) for the hashmap.

5. Top K Frequent Elements: Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.
Approach using hashmap and min heap:
    a. Create a hashmap to store the frequency of each element in nums.
    b. Iterate through the hashmap and add each element and its frequency to a min heap (priority queue) of size k. If the size of the heap exceeds k, remove the element with the lowest frequency.
    c. After iterating through the hashmap, extract the elements from the min heap and return them as the k most frequent elements.
TC:O(n*log(n)) where n is the number of unique elements in nums, SC: O(n) for the hashmap and the min heap.
