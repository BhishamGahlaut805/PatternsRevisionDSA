# PATTERN 10 Greedy and Bit Manipulation Algorithms

Greedy algorithms are a class of algorithms that make locally optimal choices at each step with the hope of finding a global optimum. They are often used to solve optimization problems, where the goal is to find the best solution among a set of feasible solutions.

1.Fractional Knapsack Problem: Given a set of items, each with a weight and a value, determine the maximum value that can be obtained by selecting items to include in a knapsack of limited capacity. In the fractional knapsack problem, you can take fractions of items.
approach using Greedy Algorithm:
    a. Calculate the value-to-weight ratio for each item.
    b. Sort the items in descending order based on their value-to-weight ratio.
    c. Initialize the total value and remaining capacity of the knapsack.
    d. Iterate through the sorted items:
        i. If the item's weight is less than or equal to the remaining capacity, add its full value to the total value and decrease the remaining capacity accordingly.
        ii. If the item's weight is greater than the remaining capacity, take a fraction of the item that fits into the knapsack, add its proportional value to the total value, and set the remaining capacity to zero.
    e. Return the total value obtained.
    TC: O(n log n) due to sorting, SC: O(1) if we don't consider the input storage.
