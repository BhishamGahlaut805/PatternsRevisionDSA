# PATTERN 6 Prefix Sum and HashSets

Prefix Sum is a technique used to calculate the sum of elements in an array or list efficiently. It involves creating a new array where each element at index `i` represents the sum of all elements from the start of the original array up to index `i`. This allows for quick calculations of subarray sums.

```python
Problems based on Prefix Sum :
1.Range Sum Query : Find the sum between 2 indices left and right in an array
approach :
    a.Create a array prefix sum storing : prefix[i]=current+prefix[i-1]
    b.if left==0 return array[right]
    c.else return array[right]-array[left-1]
TC:O(n),SC:O(n)
```

```python
2.Subarray sum equals to k : given an array nums Return the number of subarrays of nums having sum equals to k
approach:
    1.Keep a prefix sum array.
    2.Use an unordered_map {prefix sum,count}
    3.for each element p in prefix :
    4.if p ==k : then count ++
    5.if we found (p-k) in map : then ans=ans + map(p-k)
    6.increment map[p]++;
    7.return ans
TC:O(n); SC:O(n)
```

3.Find Pivot index : where left sum = right sum
approach :
    a. Calculate the total sum of the array.
    b.Initialize leftSum=0 and right Sum=totalSum
    c. for i from 1 to n:
        - rightSum = rightSum - arr[i]
        - if leftSum == rightSum:
            - return i
        - leftSum = leftSum + arr[i]
TC:O(n)

4.Product of Array Except Self
approach :
    1.Maintain left and right product initialize left=1 and right=1 and an ans array of size n
    2.for i from 0 to n:
        ans[i] =ans[i]*left
        left = left*arr[i]
    3.for i from n-1 to 0:
        ans[i] =ans[i]*right
        right = right*arr[i] 
    4.return ans
TC:O(n)

5.Number of 0 filled subarrays :
approach :
    1.Maintain a variable count=0 and ans=0
    2.for i from 0 to n:
        - if arr[i]==0:
            - count++       //Streak count of 0's
            - ans = ans + count //because if we have a streak of 0's of length k, then the number of subarrays that can be formed is k*(k+1)/2, which is equal to adding count for each 0 encountered.
        - else:
            - count=0
TC:O(n)

6.Subarray sum divisible by k :
approach :
    1.Maintain a hashmap to store the frequency of prefix sums modulo k.
    2.Initialize prefixSum=0 and count=0
    3.for i from 0 to n:
        - prefixSum = prefixSum + arr[i]
        - mod = (prefixSum % k + k) % k //to handle negative values
        - if mod in hashmap:
            - count += hashmap[mod] //add the frequency of this mod value to count
        - increment the frequency of mod in hashmap
    4.return count

    TC:O(N)
