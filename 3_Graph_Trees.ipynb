# PATTERN 3 Graph and Tree

A graph is a data structure that consists of a finite set of vertices (or nodes) and a set of edges that connect pairs of vertices. Graphs can be used to represent various real-world problems, such as social networks, transportation systems, and computer networks.

```python
Representation of Graphs :
1.Using Adjacency matrix format : TC:O(v^2) for traversal where n is the number of vertices
2.Using Adjacency list : TC of O(v+e) where v is the number of vertices in the graph and e is the number of edges.

Graph Traversals : Breadth First Traversal, Depth First Traversal
```

1.BFS approach:
    a.Use a queue to keep track of the vertices to be explored.
    b.Start from a source vertex and enqueue it.
    c.While the queue is not empty, dequeue a vertex and explore its neighbors.
    d.Mark each visited vertex to avoid revisiting it.
    e.Continue this process until all reachable vertices have been explored.
TC is O(V + E), where V is the number of vertices and E is the number of edges in the graph if using Adjacency list representation. If using adjacency matrix representation, the time complexity is O(V^2).

2.DFS approach:
    a.Use a stack (or recursion) to keep track of the vertices to be explored.
    b.Start from a source vertex and push it onto the stack.
    c.While the stack is not empty, pop a vertex and explore its neighbors.
    d.Mark each visited vertex to avoid revisiting it.
    e.Continue this process until all reachable vertices have been explored.
TC is O(V + E), where V is the number of vertices and E is the number of edges in the graph if using Adjacency list representation. If using adjacency matrix representation, the time complexity is O(V^2).

4.Single Source Shortest Path (Dijkstra's Algorithm):
    a.Initialize a distance array to keep track of the shortest distance from the source vertex to each vertex.
    b.Set the distance of the source vertex to 0 and all other vertices to infinity.
    c.Use a priority queue (min-heap) to explore vertices with the smallest distance first.
    d.For each explored vertex, update the distances of its neighbors if a shorter path is found.
    e.Continue this process until all vertices have been explored or the priority queue is empty.

    TC:O(n+ElogV) where n is the number of vertices and E is the number of edges in the graph.

5.All pairs shortest Path (Floyd-warshall ):
    a.Initialize a 2D distance array to keep track of the shortest distances between all pairs of vertices initially set to infinity, except for the diagonal elements which are set to 0.
    b.For each edge in the graph, update the distance array with the weight of the edge.
    c.For each intermediate vertex k, iterate through all pairs of vertices (i, j) and update the distance array if a shorter path is found through vertex k.
    d.Continue this process for all intermediate vertices until the distance array contains the shortest distances between all pairs of vertices.

    Fill the distance matrix first with the current distances provided by edge weights and distance of node i to itself as 0. Then for each node k, check if the distance from i to j can be minimized by going through k. If yes, update the distance from i to j.
    for each k in range(n):
        for each i in range(n):
            for each j in range(n):
                if distance[i][j] > distance[i][k] + distance[k][j]:
                    distance[i][j] = distance[i][k] + distance[k][j]
    
    TC:O(n^3) where n is the number of vertices in the graph. This is because we have three nested loops, each iterating over all vertices. The space complexity is O(n^2) for storing the distance matrix.

1.Insertion in Binary Search Tree (BST):
    a.Start from the root of the tree.
    b.Compare the value to be inserted with the current node's value.
    c.If the value is smaller, move to the left child; if larger, move to the right child.
    d.Repeat this process until reaching a null position where the new node can be inserted.
    e.Create a new node with the value and insert it at the null position.
    f. The time complexity of insertion in a BST is O(h), where h is the height of the tree. In the worst case, the height can be equal to the number of nodes (O(n)), but in a balanced BST, the height is O(log n).

2.Deletion of a Node in a BST :
    a. Start from the root of the tree.
    b. Compare the value to be deleted with the current node's value.
    c. If the value is smaller, move to the left child; if larger, move to the right child.
    d. Repeat this process until finding the node to be deleted.
    e. Once the node is found, handle its deletion based on the number of children it has:
        i. If the node has no children (leaf node), simply remove it.
        ii. If the node has one child, replace the node with its child.
        iii. If the node has two children, find the inorder successor (smallest value in the right subtree), replace the node's value with the successor's value, and then delete the successor.

3.Height of a Binary Tree:
    a. The height of a binary tree is defined as the number of edges on the longest path from the root to a leaf node.
    b. To calculate the height, you can use a recursive approach:
        i. If the current node is null, return -1 (base case).
        ii. Recursively calculate the height of the left and right subtrees.
        iii. Return the maximum of the left and right subtree heights plus one (for the current node).
    c. The time complexity of calculating the height is O(n), where n is the number of nodes in the tree, as each node is visited once.

        iv. Return the total counts of internal and external nodes.

11.Level order Traversal of a Binary Tree:
    a. Level order traversal visits all the nodes of a binary tree level by level from left to right.
    b. To perform level order traversal, we can use a queue data structure:
        i. Initialize an empty queue and enqueue the root node.
        ii. While the queue is not empty, dequeue a node and process it (e.g., print its value).
        iii. Enqueue the left and right children of the dequeued node (if they exist).
        iv. Repeat this process until all nodes have been processed.
    c. The time complexity of level order traversal is O(n), where n is the number of nodes in the tree, as each node is visited once.

12.Finding the lowest common ancestor (LCA) of two nodes in a binary tree:
    a. The LCA of two nodes p and q in a binary tree is defined as the lowest node that has both p and q as descendants (where we allow a node to be a descendant of itself).
    b. To find the LCA, you can use a recursive approach:
        i. If the current node is null, return null (base case).
        ii. If the current node is equal to either p or q, return the current node.
        iii. Recursively search for p and q in the left and right subtrees.
        iv. If both left and right recursive calls return non-null values, it means p and q are found in different subtrees, so return the current node as the LCA.
        v. If only one of the recursive calls returns a non-null value, return that value (either left or right).
    c. The time complexity of finding the LCA is O(n), where n is the number of nodes in the tree, as each node is visited once.

13.Path Sum in a Binary Tree:

    b. To solve this problem, you can use a recursive approach:
        i. If the current node is null, return (base case).
        ii. Add the current node's value to the current path and update the remaining target sum.
        iii. If the current node is a leaf and the remaining target sum equals zero, add the current path to the result list.
        iv. Recursively call the function for the left and right children of the current node.
        v. Backtrack by removing the current node's value from the current path before returning to the previous level.
    c. The time complexity of finding all path sums is O(n), where n is the number of nodes in the tree, as each node is visited once. The space complexity is O(h), where h is the height of the tree, due to the recursion stack and storage of paths.
